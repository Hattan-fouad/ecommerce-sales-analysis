```python
import pandas as pd
import numpy as np


np.random.seed(42)
n_rows = 500

data = {
    'Order_ID': range(1001, 1001 + n_rows),
    'Date': pd.date_range(start='2025-01-01', periods=n_rows, freq='h'),
    'Product': np.random.choice(['Laptop', 'Headphones', 'Smart Watch', 'Monitor', 'Keyboard'], size=n_rows),
    'Quantity': np.random.randint(1, 6, size=n_rows),
    'Unit_Price': np.random.choice([1200, 150, 300, 450, 80], size=n_rows),
    'Customer_Region': np.random.choice(['Riyadh', 'Jeddah', 'Dammam', 'Mecca'], size=n_rows)
}

df = pd.DataFrame(data)

#نضيف اخطاء متعمدين عشان ننضفها بعدين
df.loc[df.sample(frac=0.04).index, 'Quantity'] = np.nan

# نعرض اول 5 اسطر
df.head()
```

```python
# الان راح نستبدل القيم المفقودة بالوسيط
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

#الان راح نتاكد
df.isnull().sum()
```

```python
#الان بنحسب الاجمالي
df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

df.head()
```

```python
# الان بنشوف اجمالي المبيعات حسب المدن
sales_by_region = df.groupby('Customer_Region')['Total_Sales'].sum().sort_values(ascending=False)
print("إجمالي المبيعات على حسب المدينة:")
print(sales_by_region)
```

```python
#الان بنحسب اجمالي المبيعات حسب المنتج
sales_by_product = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
print("إجمالي المبيعات على حسب المنتج:")
print(sales_by_product)
```

```python
#الان بنحفظ البيانت الي نظفناها في ملف csv
df.to_csv('Cleand_Sales_Data.csv',index=False)
```
